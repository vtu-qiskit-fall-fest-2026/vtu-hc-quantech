## 2. Single-Qubit Superposition (Hadamard Gate)
Applying an $H$ gate to $|0\rangle$ yields the equal superposition state $|+\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}}$.

```python
h_circuit = {
    "num_qubits": 1,
    "operations": [{"gate": "H", "qubits": [0]}]
}

result = simulate_circuit(h_circuit, shots=1024)
print("Probabilities:", result["probabilities"])
print("Bloch Vector:", result["qubit_telemetry"][0]["bloch_vector"])
print("Is Entangled:", result["is_entangled"])
```

## 3. Two-Qubit Maximally Entangled Bell State ($|\Phi^+\rangle$)
Applying an $H$ gate on qubit 0 followed by a $CNOT$ gate between qubits 0 and 1 generates the Bell state:
$$|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}}$$

```python
bell_circuit = {
    "num_qubits": 2,
    "operations": [
        {"gate": "H", "qubits": [0]},
        {"gate": "CNOT", "qubits": [0, 1]}
    ]
}

bell_result = simulate_circuit(bell_circuit, shots=2048)
print("Bell State Measurement Probabilities:", bell_result["probabilities"])
print("Entanglement detected:", bell_result["is_entangled"])
print("Measurement Shots Summary:", bell_result.get("shots_counts"))
```

## 4. Automatic Qiskit 1.x Code Generation & Validation
Quantech provides immediate, verified export to pure Qiskit 1.x code.

```python
qiskit_code = export_to_qiskit(bell_circuit)
print("--- Generated Qiskit Code ---\n")
print(qiskit_code)

# Execute generated code to verify compatibility
local_env = {}
exec(qiskit_code, {}, local_env)
qc = local_env["qc"]
print("\nQiskit QuantumCircuit verified with depth:", qc.depth())
```

## 5. Three-Qubit GHZ State Simulation
Generating the Greenberger–Horne–Zeilinger (GHZ) state $|GHZ\rangle = \frac{|000\rangle + |111\rangle}{\sqrt{2}}$.

```python
ghz_circuit = {
    "num_qubits": 3,
    "operations": [
        {"gate": "H", "qubits": [0]},
        {"gate": "CNOT", "qubits": [0, 1]},
        {"gate": "CNOT", "qubits": [1, 2]}
    ]
}

ghz_result = simulate_circuit(ghz_circuit, shots=1000)
print("GHZ State Probabilities:", ghz_result["probabilities"])
print("Entanglement Detected:", ghz_result["is_entangled"])
```
